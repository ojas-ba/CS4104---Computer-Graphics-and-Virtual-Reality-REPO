### Aim
To implement the Midpoint Circle Drawing Algorithm using eight-way symmetry.

```python
%%writefile experiment2.cpp
#include <iostream>
using namespace std;

void plot8(int xc, int yc, int x, int y, const char* tag) {
    cout << tag << " " << xc+x << " " << yc+y << "\n";
    cout << tag << " " << xc-x << " " << yc+y << "\n";
    cout << tag << " " << xc+x << " " << yc-y << "\n";
    cout << tag << " " << xc-x << " " << yc-y << "\n";
    cout << tag << " " << xc+y << " " << yc+x << "\n";
    cout << tag << " " << xc-y << " " << yc+x << "\n";
    cout << tag << " " << xc+y << " " << yc-x << "\n";
    cout << tag << " " << xc-y << " " << yc-x << "\n";
}

void midpointCircle(int xc, int yc, int r, const char* tag) {
    int x = 0, y = r;
    int p = 1 - r;

    while (x <= y) {
        plot8(xc, yc, x, y, tag);
        x++;
        if (p < 0)
            p += 2 * x + 1;
        else {
            y--;
            p += 2 * (x - y) + 1;
        }
    }
}

int main() {
    int xc = 450, yc = 350;
    midpointCircle(xc, yc, 250, "OUTER");
    midpointCircle(xc, yc, 165, "MIDDLE");
    midpointCircle(xc, yc, 85,  "INNER");
    return 0;
}
```

```python
!g++ experiment2.cpp -o experiment2
output = !./experiment2

points = {'OUTER': [], 'MIDDLE': [], 'INNER': []}
for line in output:
    tag, x, y = line.split()
    points[tag].append((int(x), int(y)))

print('Midpoint Circle Algorithm executed successfully.')
```

```python
from PIL import Image, ImageDraw
from IPython.display import display

W, H = 900, 700
img = Image.new('RGB', (W, H), (23, 31, 40))
draw = ImageDraw.Draw(img)

colors = {
    'OUTER':  (0, 245, 245),
    'MIDDLE': (255, 235, 0),
    'INNER':  (255, 20, 45)
}

for tag, pts in points.items():
    for x, y in pts:
        sy = H - y
        draw.rectangle((x-2, sy-2, x+2, sy+2), fill=colors[tag])

img.save('experiment2_midpoint_circle.png')
display(img)
print('Saved as experiment2_midpoint_circle.png')
```
